## Objective
Analyze banking transaction data to understand transaction patterns, transaction amounts, channels, account types, and transaction status.

**Dataset note:** The dataset is synthetic and created for educational purposes. It contains no real customer or financial information.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("bank_transactions.csv")
df["Date"] = pd.to_datetime(df["Date"])
print("Dataset shape:", df.shape)
df.head()
```

## 1. Data Understanding

```python
print(df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
df["Amount"].describe()
```

## 2. Data Cleaning

```python
df = df.drop_duplicates().copy()
df = df[df["Amount"] > 0].copy()
print("Rows after cleaning:", len(df))
print(df.isnull().sum())
```

## 3. Transaction-Type Analysis

```python
type_summary = df.groupby("Transaction_Type").agg(
    Transaction_Count=("Transaction_ID","count"),
    Total_Amount=("Amount","sum"),
    Average_Amount=("Amount","mean")
).sort_values("Total_Amount", ascending=False)
type_summary
```

```python
type_summary["Transaction_Count"].plot(kind="bar", title="Transactions by Type")
plt.xlabel("Transaction Type"); plt.ylabel("Count"); plt.xticks(rotation=0)
plt.tight_layout(); plt.show()
```

## 4. Channel Analysis

```python
channel_summary = df.groupby("Channel").agg(
    Transaction_Count=("Transaction_ID","count"),
    Total_Amount=("Amount","sum")
).sort_values("Transaction_Count", ascending=False)
channel_summary
```

```python
channel_summary["Transaction_Count"].plot(kind="bar", title="Transactions by Channel")
plt.xlabel("Channel"); plt.ylabel("Count"); plt.xticks(rotation=30)
plt.tight_layout(); plt.show()
```

## 5. Account-Type Analysis

```python
account_summary = df.groupby("Account_Type").agg(
    Transaction_Count=("Transaction_ID","count"),
    Total_Amount=("Amount","sum"),
    Average_Amount=("Amount","mean")
)
account_summary
```

## 6. Monthly Transaction Trend

```python
monthly = df.set_index("Date").resample("ME").agg(
    Transaction_Count=("Transaction_ID","count"),
    Total_Amount=("Amount","sum")
)
monthly
```

```python
monthly["Transaction_Count"].plot(kind="line", marker="o", title="Monthly Transaction Count")
plt.xlabel("Month"); plt.ylabel("Transaction Count"); plt.grid(True)
plt.tight_layout(); plt.show()
```

## 7. Transaction Status

```python
status_summary = df["Status"].value_counts()
print(status_summary)
status_summary.plot(kind="pie", autopct="%1.1f%%", title="Transaction Status")
plt.ylabel(""); plt.tight_layout(); plt.show()
```

## 8. Key Findings

```python
print(f"Total transactions: {len(df):,}")
print(f"Total transaction value: {df['Amount'].sum():,.2f}")
print(f"Average transaction amount: {df['Amount'].mean():,.2f}")
print(f"Successful transaction rate: {df['Status'].eq('Success').mean()*100:.2f}%")
print("Most common transaction type:", df["Transaction_Type"].mode().iloc[0])
print("Most used channel:", df["Channel"].mode().iloc[0])
```
